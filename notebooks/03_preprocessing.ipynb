# 🧹 Dataset Preprocessing

## Objective

This notebook prepares the augmented dataset for model training.

The preprocessing steps remove common formatting issues while preserving the original meaning of each sentence.

The output of this notebook will be used for dataset splitting and mBART-50 fine-tuning.

---

**Input**

- augmented_dataset.xlsx

**Output**

- preprocessed_dataset.xlsx

---

```python
# ==========================================================
# Import Required Libraries
# ==========================================================

# pandas is used for loading and processing the dataset.
import pandas as pd

# re is used to clean unnecessary whitespace.
import re
```

```python
# ==========================================================
# Load Augmented Dataset
# ==========================================================

# Load the augmented dataset generated
# from the previous notebook.

df = pd.read_excel("augmented_dataset.xlsx")

df.columns = ["Bangla", "Chatgaya"]

print(f"Total Sentence Pairs : {len(df)}")

df.head()
```

```python
# ==========================================================
# Remove Missing Values
# ==========================================================

# Remove rows where either Bangla or
# Chatgaya sentence is missing.

before = len(df)

df = df.dropna()

after = len(df)

print(f"Removed Rows : {before-after}")

print(f"Remaining Rows : {after}")
```

```python
# ==========================================================
# Remove Leading and Trailing Spaces
# ==========================================================

# Remove unnecessary spaces from the
# beginning and end of each sentence.

df["Bangla"] = df["Bangla"].astype(str).str.strip()

df["Chatgaya"] = df["Chatgaya"].astype(str).str.strip()
```

```python
# ==========================================================
# Normalize Multiple Spaces
# ==========================================================

# Replace multiple consecutive spaces
# with a single space.

def normalize_spaces(text):

    text = re.sub(r"\s+", " ", text)

    return text.strip()


df["Bangla"] = df["Bangla"].apply(normalize_spaces)

df["Chatgaya"] = df["Chatgaya"].apply(normalize_spaces)
```

```python
# ==========================================================
# Remove Duplicate Sentence Pairs
# ==========================================================

# Remove any duplicate sentence pairs.

before = len(df)

df = df.drop_duplicates().reset_index(drop=True)

after = len(df)

print(f"Duplicate Pairs Removed : {before-after}")

print(f"Remaining Sentence Pairs : {after}")
```

```python
# ==========================================================
# Sentence Length Statistics
# ==========================================================

# Calculate sentence length statistics.
# This information helps determine
# the maximum sequence length for training.

bangla_lengths = df["Bangla"].apply(lambda x: len(x.split()))

chatgaya_lengths = df["Chatgaya"].apply(lambda x: len(x.split()))

print("="*60)
print("SENTENCE STATISTICS")
print("="*60)

print(f"Bangla Average Length   : {bangla_lengths.mean():.2f}")
print(f"Bangla Maximum Length   : {bangla_lengths.max()}")

print()

print(f"Chatgaya Average Length : {chatgaya_lengths.mean():.2f}")
print(f"Chatgaya Maximum Length : {chatgaya_lengths.max()}")
```

```python
# ==========================================================
# Save Preprocessed Dataset
# ==========================================================

OUTPUT_FILE = "preprocessed_dataset.xlsx"

df.to_excel(
    OUTPUT_FILE,
    index=False
)

print(f"Dataset saved as '{OUTPUT_FILE}'")
```
