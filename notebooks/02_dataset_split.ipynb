```python
# ==========================================================
# Import Required Libraries
# ==========================================================

# pandas is used to load and save datasets.
import pandas as pd

# train_test_split is used to divide the dataset.
from sklearn.model_selection import train_test_split
```

```python
# ==========================================================
# Load Canonical Dataset
# ==========================================================

# Load the manually created canonical dataset.

df = pd.read_excel("canonical.xlsx")

df.columns = ["Bangla", "Chatgaya"]

print(f"Total Sentence Pairs : {len(df)}")

df.head()
```

## Split Strategy

The dataset is divided using an **80/10/10** ratio.

- 80% → Training
- 10% → Validation
- 10% → Test

A fixed random seed is used to ensure that the split is reproducible.

```python
# ==========================================================
# Dataset Split
# ==========================================================

# First split:
# 80% Train
# 20% Temporary

train_df, temp_df = train_test_split(
    df,
    test_size=0.20,
    random_state=42,
    shuffle=True
)

# Second split:
# 10% Validation
# 10% Test

validation_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42,
    shuffle=True
)

print("="*60)
print("DATASET SPLIT")
print("="*60)

print(f"Training Set   : {len(train_df)}")

print(f"Validation Set : {len(validation_df)}")

print(f"Test Set       : {len(test_df)}")
```

## Verify Dataset Independence

To ensure a fair evaluation, the training, validation, and test sets must be completely independent.

This section checks whether any identical sentence pairs exist across different splits.

If the overlap count is zero, the dataset split is considered valid.

```python
# ==========================================================
# Check Overlap Between Splits
# ==========================================================

# Convert each dataset into a set of sentence pairs.
# This allows us to efficiently compare different splits.

train_pairs = set(
    zip(train_df["Bangla"], train_df["Chatgaya"])
)

validation_pairs = set(
    zip(validation_df["Bangla"], validation_df["Chatgaya"])
)

test_pairs = set(
    zip(test_df["Bangla"], test_df["Chatgaya"])
)

# Calculate overlaps

train_validation_overlap = train_pairs.intersection(validation_pairs)

train_test_overlap = train_pairs.intersection(test_pairs)

validation_test_overlap = validation_pairs.intersection(test_pairs)

print("=" * 60)
print("OVERLAP CHECK")
print("=" * 60)

print(f"Train ↔ Validation : {len(train_validation_overlap)}")

print(f"Train ↔ Test       : {len(train_test_overlap)}")

print(f"Validation ↔ Test  : {len(validation_test_overlap)}")
```

## Save Split Datasets

The three datasets are saved separately.

Only the training set will be used for augmentation in the next notebook.

The validation and test sets will remain unchanged to ensure an unbiased evaluation.

```python
# ==========================================================
# Save Dataset Splits
# ==========================================================

train_df.to_excel(
    "train.xlsx",
    index=False
)

validation_df.to_excel(
    "validation.xlsx",
    index=False
)

test_df.to_excel(
    "test.xlsx",
    index=False
)

print("All datasets saved successfully.")
```
