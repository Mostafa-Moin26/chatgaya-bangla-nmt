# 🔄 Training Dataset Augmentation

## Objective

To improve the diversity of the training data, lexical augmentation is applied **only to the training dataset**.

The validation and test datasets are intentionally left unchanged to ensure a fair and unbiased evaluation.

This notebook generates additional training sentence pairs by replacing Chatgaya lexical variants while preserving the original Bangla sentences.

---

**Input**

- train.xlsx
- lexicon.xlsx

**Output**

- train_augmented.xlsx

```python
# ==========================================================
# Import Required Libraries
# ==========================================================

# pandas is used for reading and writing Excel files.
import pandas as pd

# itertools will be used to generate different
# combinations of lexical replacements.
import itertools
```

```python
# ==========================================================
# Load Training Dataset
# ==========================================================

# Load only the training dataset.
# Validation and test datasets are not augmented.

train_df = pd.read_excel("train.xlsx")
train_df.columns = ["Bangla", "Chatgaya"]

lexicon_df = pd.read_excel("lexicon.xlsx")
lexicon_df.columns = ["Meaning", "Bangla", "Chatgaya"]

print(f"Training Sentence Pairs : {len(train_df)}")
print(f"Lexicon Entries         : {len(lexicon_df)}")
```

```python
# ==========================================================
# Preview Dataset
# ==========================================================

print("Training Dataset")
display(train_df.head())

print("\nLexicon Dataset")
display(lexicon_df.head())
```

# Build Chatgaya Lexicon

## Objective

The lexicon contains one or more Chatgaya variations for each Bangla word.

In this section, we convert the lexicon into a Python dictionary so that each Bangla word can quickly retrieve all of its Chatgaya variations.

Example:

আজকে

↓

["আজিয়ে", "আজিইয়্যে"]

```python
# ==========================================================
# Build Lexicon Dictionary
# ==========================================================

# Convert the lexicon into a dictionary.
#
# Example:
#
# {
#     "আজকে": ["আজিয়ে", "আজিইয়া"],
#     "সকালে": ["ফজরাত", "বেইন্নে"]
# }

lexicon = {}

for _, row in lexicon_df.iterrows():

    bangla_words = [
        word.strip()
        for word in str(row["Bangla"]).split("|")
    ]

    chatgaya_words = [
        word.strip()
        for word in str(row["Chatgaya"]).split("|")
    ]

    for bangla_word in bangla_words:

        lexicon[bangla_word] = chatgaya_words

print(f"Total Lexicon Entries : {len(lexicon)}")
```

```python
print(lexicon)
```

```python
# ==========================================================
# Preview Lexicon
# ==========================================================

# Display the first few entries
# to verify that the dictionary
# has been created correctly.

count = 0

for key, value in lexicon.items():

    print(f"{key} -> {value}")

    count += 1

    if count == 10:
        break
```

# Generate Augmented Sentence

## Strategy

For each sentence pair,

1. Keep the original sentence.
2. Detect the Chatgaya variation already used.
3. Replace only that Chatgaya word with the other available variations.
4. Keep the Bangla sentence unchanged.

This produces high-quality lexical augmentation while preserving sentence meaning.

```python
# ==========================================================
# Sentence Augmentation Function
# ==========================================================

# This function generates augmented sentence pairs
# by replacing only one Chatgaya variation at a time.
#
# Bangla sentence remains unchanged.
# Only the Chatgaya sentence is modified.

def augment_sentence(bangla_sentence, chatgaya_sentence):

    augmented_pairs = []

    # Keep the original sentence
    augmented_pairs.append((bangla_sentence, chatgaya_sentence))

    # Split Chatgaya sentence into words
    chatgaya_words = chatgaya_sentence.split()

    # Check every Bangla word in the lexicon
    for bangla_word, variations in lexicon.items():

        # Skip if the Bangla word is not present
        # in the Bangla sentence
        if bangla_word not in bangla_sentence:
            continue

        # Find which variation is currently used
        current_variation = None

        for variation in variations:

            if variation in chatgaya_words:
                current_variation = variation
                break

        # Skip if none of the variations exist
        if current_variation is None:
            continue

        # Generate one sentence for every
        # remaining variation
        for new_variation in variations:

            # Skip the original variation
            if new_variation == current_variation:
                continue

            # Copy the sentence words
            new_words = chatgaya_words.copy()

            # Replace only one word
            for i, word in enumerate(new_words):

                if word == current_variation:

                    new_words[i] = new_variation
                    break

            # Join back into a sentence
            new_chatgaya = " ".join(new_words)

            augmented_pairs.append(
                (
                    bangla_sentence,
                    new_chatgaya
                )
            )

    return augmented_pairs
```

```python
# ==========================================================
# Generate Augmented Dataset
# ==========================================================

# Apply augmentation to every sentence pair
# in the training dataset.

augmented_pairs = []

for _, row in train_df.iterrows():

    pairs = augment_sentence(
        row["Bangla"],
        row["Chatgaya"]
    )

    augmented_pairs.extend(pairs)

print(f"Generated Sentence Pairs : {len(augmented_pairs)}")
```

```python
# ==========================================================
# Convert into DataFrame
# ==========================================================

augmented_df = pd.DataFrame(
    augmented_pairs,
    columns=["Bangla", "Chatgaya"]
)

print(f"Before Removing Duplicates : {len(augmented_df)}")
```

```python
# ==========================================================
# Remove Duplicate Sentence Pairs
# ==========================================================

before = len(augmented_df)

augmented_df = augmented_df.drop_duplicates().reset_index(drop=True)

after = len(augmented_df)

print(f"Duplicates Removed : {before-after}")

print(f"Final Dataset Size : {after}")
```
